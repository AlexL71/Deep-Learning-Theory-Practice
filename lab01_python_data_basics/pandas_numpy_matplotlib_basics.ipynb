# Pandas 핵심 데이터 구조  

Pandas는 **라벨이 붙은 표 형식 데이터**를 다루기 위한 Python 라이브러리입니다.  
데이터 구조(Data Structures)는 크게 3가지가 있었지만, 오늘은 **2가지만에 대해서만 다뤄본다**:

- **Series (시리즈)**: 1차원, 라벨이 있는 배열 (1D labeled array)
- **DataFrame (데이터프레임)**: 2차원, 라벨이 있는 표 (2D labeled table)
- **Panel (패널)**

> 체크: **Panel은 끝났습니다.** 최신 Pandas에서는 지원하지 않습니다.

---

## 언제 무엇을 쓰나?

- **Series**: 한 열(column) 또는 단일 측정값 목록이 필요할 때  
  예) 키 리스트, 한 종목의 주가 시계열
- **DataFrame**: 행-열로 이루어진 **표 전체**를 다룰 때 (가장 많이 사용)  
  예) CSV/Excel 전체, 학습 데이터셋 테이블

---

### 정의

# 시리즈(Series)

시리즈 클래스는 1차원 배열의 값(values)에 각 값에 대응되는 인덱스(index)를 부여할 수 있는 구조

```python
sr = pd.Series([17000, 18000, 1000, 5000],
               index=["피자", "치킨", "콜라", "맥주"])
print('시리즈 출력 :')
print('-'*15)
print(sr)
```

```python
sr
```

```python
print('시리즈의 값 : {}'.format(sr.values))
print('시리즈의 인덱스 : {}'.format(sr.index))
```

```python
sr.values
```

```python
sr.index
```

```python
sr.dtype
```

# 데이터프레임(DataFrame)

데이터프레임은 2차원 리스트를 매개변수로 전달한다. 

2차원이므로 행방향 인덱스(index)와 열방향 인덱스(column)가 존재한다. 

다시 말해 행과 열을 가지는 자료구조이다. 

시리즈가 인덱스(index)와 값(values)으로 구성된다면, 데이터프레임은 열(columns)까지 추가되어 열(columns), 인덱스(index), 값(values)으로 구성된다.

```python
values = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
index = ['one', 'two', 'three']
columns = ['A', 'B', 'C']
```

```python
df = pd.DataFrame(values, index=index, columns=columns)
```

```python
print('데이터프레임 출력 :')
print('-'*18)
print(df)
```

```python
df
```

```python
print('데이터프레임의 인덱스 : {}'.format(df.index))
print('데이터프레임의 열이름: {}'.format(df.columns))
print('데이터프레임의 값 :')
print('-'*18)
print(df.values)
```

```python
df.columns=['가','나','다']
```

```python
df
```

```python
df.iloc[:,:]
```

```python
df.iloc[2:,1:2]
```

```python
import numpy as np
```

Numpy의 핵심은 ndarray이다. np.array()는 리스트, 튜플, 배열로 부터 ndarray를 생성한다. 

```python
#1차원 배열
vec = np.array([1, 2, 3, 4, 5])
print(vec)
```

```python
# 2차원 배열
mat = np.array([[10, 20, 30], [ 60, 70, 80]]) 
print(mat)
```

```python
print('vec의 타입 :',type(vec))
print('mat의 타입 :',type(mat))
```

Numpy 배열에는 축의 개수(ndim)와 크기(shape)라는 개념이 존재하는데, 배열의 크기를 정확히 숙지하는 것은 딥 러닝에서 매우 중요

```python
print('vec의 축의 개수 :',vec.ndim) # 축의 개수 출력
print('vec의 크기(shape) :',vec.shape) # 크기 출력
```

```python
print('mat의 축의 개수 :',mat.ndim) # 축의 개수 출력
print('mat의 크기(shape) :',mat.shape) # 크기 출력
```

```python
mat.reshape(2,3,1)
```

ndarray의 초기화

```python
# 모든 값이 0인 2x3 배열 생성.
zero_mat = np.zeros((2,3))
print(zero_mat)
```

```python
# 모든 값이 1인 2x3 배열 생성.
one_mat = np.ones((2,3))
print(one_mat)
```

```python
# 모든 값이 특정 상수인 배열 생성. 이 경우 7.
same_value_mat = np.full((2,2), 7)
print(same_value_mat)
```

np.eye()는 대각선으로는 1이고 나머지는 0인 2차원 배열을 생성

```python
# 대각선 값이 1이고 나머지 값이 0인 2차원 배열을 생성.
eye_mat = np.eye(3)
print(eye_mat)
```

np.random.random()은 임의의 값을 가지는 배열을 생성

```python
# 임의의 값으로 채워진 배열 생성
random_mat = np.random.random((2,2)) # 임의의 값으로 채워진 배열 생성
print(random_mat)
```

np.arange(n)은 0부터 n-1까지의 값을 가지는 배열을 생성

```python
# 0부터 9까지
range_vec = np.arange(10)
print(range_vec)
```

np.arange(i, j, k)는 i부터 j-1까지 k씩 증가하는 배열을 생성

```python
# 1부터 9까지 +2씩 적용되는 범위
n = 2
range_n_step_vec = np.arange(1, 10, n)
print(range_n_step_vec)
```

np.reshape()은 내부 데이터는 변경하지 않으면서 배열의 구조를 바꾼다.

0부터 29까지의 숫자를 생성하는 arange(30)을 수행한 후, 원소의 개수가 30개이므로 5행 6열의 행렬로 변경

```python
reshape_mat = np.array(np.arange(30)).reshape((5,6))
print(reshape_mat)
```

ndarray를 통해 만든 다차원 배열은 파이썬의 자료구조인 리스트처럼 슬라이싱(slicing) 기능을 지원한다. 

슬라이싱 기능을 사용하여 특정 행이나 열들의 원소들을 접근할 수 있다.

```python
mat = np.array([[1, 2, 3], [4, 5, 6]])
print(mat)
```

```python
# 첫번째 행 출력
slicing_mat = mat[0, :]  # pandas dataframe에도 동일하게 적용 가능.
print(slicing_mat)
```

```python
# 두번째 열 출력
slicing_mat = mat[:, 1]
print(slicing_mat)
```

Numpy 연산

Numpy를 사용하면 배열간 연산을 손쉽게 수행할 수 있다. 덧셈, 뺄셈, 곱셈, 나눗셈을 위해서는 연산자 +, -, *, /를 사용할 수 있으며 또는 np.add(), np.subtract(), np.multiply(), np.divide()를 사용할 수도 있다.

```python
x = np.array([1,2,3])
y = np.array([4,5,6])
```

```python
y
```

```python
# result = np.add(x, y)와 동일.
result = x + y
print(result)
```

```python
# result = np.subtract(x, y)와 동일.
result = x - y
print(result)
```

```python
# result = np.multiply(result, x)와 동일.
result = result * x
print(result)
```

```python
# result = np.divide(result, x)와 동일.
result = result / x
print(result)
```

```python
mat1 = np.array([[1,2],[3,4]])
mat2 = np.array([[5,6],[7,8]])
mat3 = np.dot(mat1, mat2)
print(mat3)
```

# 맷플롯립(Matplotlib)

맷플롯립(Matplotlib)은 데이터를 차트(chart)나 플롯(plot)으로 시각화하는 패키지이다. 

데이터 분석에서 Matplotlib은 데이터 분석 이전에 데이터 이해를 위한 시각화나, 데이터 분석 후에 결과를 시각화하기 위해서 사용된다. 

```python
pip install matplotlib
```

```python
import matplotlib.pyplot as plt
```

1) 라인 플롯 그리기
plot()은 라인 플롯을 그리는 기능을 수행한다. 

plot()에 x축과 y축의 값을 기재하고 그림을 표시하는 show()를 통해서 시각화해본다.

```python
plt.title('test') #제목 지정 
plt.plot([1,2,3,4],[2,4,8,6]) # x축, y 축 지정
plt.show()
```

2) 축 레이블 삽입하기

x축과 y축 각각에 축이름을 삽입하고 싶다면 xlabel('넣고 싶은 축이름')과 ylabel('넣고 싶은 축이름')을 사용

```python
plt.title('test')
plt.plot([1,2,3,4],[2,4,8,6])
plt.xlabel('hours')
plt.ylabel('score')
plt.show()
```

3) 라인 추가와 범례 삽입하기

다수의 plot()을 하나의 그래프에 나타낼 수 있다. 

여러개의 라인 플롯을 동시에 사용할 경우에는 각 선이 어떤 데이터를 나타내는지를 보여주기 위해 범례(legend)를 사용할 수 있다

```python
plt.title('students')
plt.plot([1,2,3,4],[2,4,8,6])
plt.plot([1.5,2.5,3.5,4.5],[3,5,8,10]) # 라인 새로 추가
plt.xlabel('hours')
plt.ylabel('score')
plt.legend(['A student', 'B student']) # 범례 삽입
plt.show()
```
